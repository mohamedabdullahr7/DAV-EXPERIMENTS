```python
import pandas as pd

# Read datasets
text_df = pd.read_csv("Google_data (2b.c1).csv")
excel_df = pd.read_excel("data (2c2).xlsx")
product_df = pd.read_csv("products.csv")

# Display data
print("Review Dataset")
print(text_df.head())

print("\nFeature Dataset")
print(excel_df.head())

print("\nProduct Dataset")
print(product_df.head())

# Handle missing values
text_df.ffill(inplace=True)
excel_df.bfill(inplace=True)
product_df.dropna(inplace=True)

# Merge datasets
merged = text_df.merge(excel_df, on="user_id")
merged = merged.merge(product_df, on="product_id")

print("\nMerged Dataset")
print(merged.head())

# Save processed data
merged.to_csv("processed_dataset.csv", index=False)

print("\nDataset successfully processed.")
```
